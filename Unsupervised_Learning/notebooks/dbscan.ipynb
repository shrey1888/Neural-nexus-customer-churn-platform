# DBSCAN Customer Segmentation
This notebook explores DBSCAN clustering on the customer dataset.

```python
import pandas as pd
import numpy as np
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('../../data/processed/cleaned_dataset.csv')
X_raw = df.drop(columns=['Churn', 'Cluster'], errors='ignore')
cat_cols = X_raw.select_dtypes(exclude='number').columns.tolist()
X_enc = pd.get_dummies(X_raw, columns=cat_cols, drop_first=True)
X_scaled = StandardScaler().fit_transform(X_enc)

db = DBSCAN(eps=3.0, min_samples=5)
db_labels = db.fit_predict(X_scaled)
print("Unique DBSCAN labels (including noise -1):", np.unique(db_labels))
print("Noise count:", sum(db_labels == -1))
```

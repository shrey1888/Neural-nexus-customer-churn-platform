# PCA Dimensionality Reduction
This notebook visualizes the high-dimensional customer profiles in 2D using Principal Component Analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('../../data/processed/cleaned_dataset.csv')
X_raw = df.drop(columns=['Churn', 'Cluster'], errors='ignore')
cat_cols = X_raw.select_dtypes(exclude='number').columns.tolist()
X_enc = pd.get_dummies(X_raw, columns=cat_cols, drop_first=True)
X_scaled = StandardScaler().fit_transform(X_enc)

pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)
print("Explained variance ratio:", pca.explained_variance_ratio_)

plt.figure(figsize=(8, 6))
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=df['Cluster'] if 'Cluster' in df.columns else None, palette='Set1', alpha=0.5)
plt.title('PCA 2D Projection')
plt.show()
```

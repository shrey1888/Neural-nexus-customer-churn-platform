# Feature Engineering - Telecom Customer Churn
In this notebook we extract business-driven features like TenureGroup, ServiceCount, and ChargesRatio.

```python
import pandas as pd
import numpy as np
import os
```

## 1. Load Data

```python
df = pd.read_csv('../../data/processed/cleaned_dataset.csv')
df.head()
```

## 2. Engineering Features

```python
# Group tenure
df['TenureGroup'] = pd.cut(df['tenure'], bins=[-1, 12, 24, 48, 60, np.inf], labels=['0-12m', '12-24m', '24-48m', '48-60m', '60m+']).astype(str)

# Count active services
services = ['PhoneService', 'MultipleLines', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies']
df['ServiceCount'] = sum((df[col] == 'Yes').astype(int) for col in services if col in df.columns)
if 'InternetService' in df.columns:
    df['ServiceCount'] += (df['InternetService'] != 'No').astype(int)

# Charge ratio
df['ChargesRatio'] = df['MonthlyCharges'] / (df['TotalCharges'] + 1e-5)

df[['tenure', 'TenureGroup', 'ServiceCount', 'ChargesRatio']].head()
```

## 3. Save Features

```python
df.to_csv('../../data/processed/cleaned_dataset.csv', index=False)
print("Successfully saved engineered features to cleaned_dataset.csv")
```

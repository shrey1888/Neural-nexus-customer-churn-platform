# Data Cleaning - Telecom Customer Churn
This notebook details the data cleaning process for the Telco Customer Churn dataset.

```python
import pandas as pd
import numpy as np
import os
```

## 1. Load Raw Data

```python
df = pd.read_csv('../../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv')
print("Raw shape:", df.shape)
df.head()
```

## 2. Inspect Column Types & Missing Values

```python
print(df.info())
print("\nMissing values:")
print(df.isnull().sum())
```

## 3. Coerce TotalCharges to Float and Impute Missing Values

```python
# TotalCharges has blank spaces
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'].replace(' ', np.nan, regex=False))
print("Nulls in TotalCharges after coercion:", df['TotalCharges'].isnull().sum())

# Fill nulls with 0.0 (these are new customers with tenure = 0)
df['TotalCharges'] = df['TotalCharges'].fillna(0.0)
print("Nulls left in TotalCharges:", df['TotalCharges'].isnull().sum())
```

## 4. Drop Unnecessary Columns and Save

```python
df = df.drop(columns=['customerID'], errors='ignore')
os.makedirs('../../data/processed', exist_ok=True)
df.to_csv('../../data/processed/cleaned_dataset.csv', index=False)
print("Saved cleaned data. Shape:", df.shape)
```

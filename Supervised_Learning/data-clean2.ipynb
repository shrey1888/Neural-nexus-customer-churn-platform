```python


import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import seaborn as sns

import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/kaggle/input/datasets/shreyy45/telecom-ds/telecom_customer_churn_synthetic.csv')
df.shape
df.info
# df.describe()
df.isnull().sum()[df.isnull().sum() > 0]/ len(df) * 100
# (df.isnull().sum() / len(df)) * 100
```

```python
cols = [
    "IncomeGroup",
    "Education",
    "OutstandingBalance",
    "CustomerSatisfactionScore",
    "AvgInternetSpeedMbps"
]

df[cols].hist(figsize=(12, 8), bins=20)
plt.tight_layout()
plt.show()
```

```python
df["OutstandingBalance"]=df["OutstandingBalance"].fillna(df["OutstandingBalance"].median())

df["AvgInternetSpeedMbps"]=df["AvgInternetSpeedMbps"].fillna(df["AvgInternetSpeedMbps"].median())

df["CustomerSatisfactionScore"]=df["CustomerSatisfactionScore"].fillna(df["CustomerSatisfactionScore"].mode()[0])
df.isnull().sum()[df.isnull().sum() > 0]/ len(df) * 100
```

```python
# df["Education"].value_counts()
# df["IncomeGroup"].value_counts()
df["Education"] = df["Education"].fillna(df["Education"].mode().iat[0])
# df["IncomeGroup"] = df["IncomeGroup"].fillna(df["IncomeGroup"].mode().iloc[0])
```
